```python
yearly = df.groupby("Financial_Year").agg({
    "Sales_Cr": "sum",
    "Expenses_Cr": "sum",
    "Operating_Profit_Cr": "sum",
    "Net_Profit_Cr": "sum"
})

yearly
```

```python
yearly["Revenue_Growth_%"] = (
    yearly["Sales_Cr"].pct_change() * 100
)

yearly["Expense_Growth_%"] = (
    yearly["Expenses_Cr"].pct_change() * 100
)

yearly["Operating_Profit_Growth_%"] = (
    yearly["Operating_Profit_Cr"].pct_change() * 100
)

yearly["Net_Profit_Growth_%"] = (
    yearly["Net_Profit_Cr"].pct_change() * 100
)

yearly
```
